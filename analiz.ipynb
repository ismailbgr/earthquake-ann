```python
import pandas as pd 
from glob import glob
#init pycaret
from pycaret.classification import *
```

```python
from tqdm import tqdm
files = glob("data/*.csv")
```

```python
main_df = pd.DataFrame()
for fil in tqdm(files, desc="Processing files", unit="file"):
    df = pd.read_csv(fil)
    df = df[["x","y","z","stresses_full_xx","stresses_full_xy","stresses_full_xz","stresses_full_yy","stresses_full_yz","stresses_full_zz","grid_aftershock_count","is_aftershock","Magnitudes_avg"]]
    main_df = pd.concat([main_df, df], ignore_index=True)
```

```python
main_df
```

```python
main_df.to_csv("combined.csv", index=False)
```

```python


#init pycaret
exp_clf = setup(data=main_df, target='is_aftershock', session_id=123,  experiment_name="aftershock", use_gpu=True)

#compare models
best = compare_models(sort='F1', n_select=3)
print(best)

```
